```python
from Geometry.Device import Device
```

```python
my_t9_hall = Device.open_file('wcte_t9.geo')
my_wcte_t9 = my_t9_hall.wcds[0]

my_bldg157_hall = Device.open_file('wcte_bldg157.geo')
my_wcte_bldg157 = my_bldg157_hall.wcds[0]
```

```python
mpmt_ids = [37,38,44]

for hall,wcte in zip(['T9','Bldg 157'],[my_wcte_t9, my_wcte_bldg157]):
    print(f"mPMT locations in {hall}:")
    for mpmt_id in mpmt_ids:
        loc = wcte.mpmts[mpmt_id].get_placement('est')['location']
        print(f"  mPMT {mpmt_id}: x={loc[0]:.2f}, y={loc[1]:.2f}, z={loc[2]:.2f}")
```
